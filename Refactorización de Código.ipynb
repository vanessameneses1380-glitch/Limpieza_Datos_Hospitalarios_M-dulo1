```python
import pandas as pd
import numpy as np
import os


def add_columns(column_a, column_b):
    """
    Add two pandas Series or arrays element-wise.
    
    Parameters:
    -----------
    column_a : pandas.Series or array-like
        First column to add
    column_b : pandas.Series or array-like
        Second column to add
    
    Returns:
    --------
    pandas.Series or array-like
        Element-wise sum of the two input columns
    """
    return column_a + column_b


def load_data_safely(file_path):
    """
    Safely load CSV data with error handling.
    
    Parameters:
    -----------
    file_path : str
        Path to the CSV file
    
    Returns:
    --------
    pandas.DataFrame or None
        Loaded dataframe or None if file not found
    """
    try:
        return pd.read_csv(file_path)
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        print("Creating sample data for demonstration...")
        # Create sample data if file doesn't exist
        sample_data = {
            'col1': [1, 2, 3, 4, 5],
            'col2': [10, 20, 30, 40, 50],
            'col3': [100, 200, 300, 400, 500],
            'col4': [5, 15, 25, 35, 45]
        }
        return pd.DataFrame(sample_data)


# Load the data safely
dataframe = load_data_safely("datos.csv")

if dataframe is not None:
    # Calculate sum of col1 and col2
    first_column = dataframe['col1']
    second_column = dataframe['col2']
    sum_result = add_columns(first_column, second_column)
    dataframe['res'] = sum_result
    
    # Calculate sum of col3 and col4 using the same function
    sum_result_2 = add_columns(dataframe['col3'], dataframe['col4'])
    dataframe['res2'] = sum_result_2
    
    # Display the first few rows
    print("Processed data:")
    print(dataframe.head())
else:
    print("Could not load or create data.")
```
